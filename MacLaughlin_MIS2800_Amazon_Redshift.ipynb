2. Speed to Insight: How does the architecture of a "Leader Node" and "Compute Nodes" change how quickly a Business Intelligence report is generated?  https://docs.aws.amazon.com/redshift/latest/dg/c_high_level_system_architecture.html


## YOUR ANSWERS TO PART 2 BELOW

Question 1:In Amazon Redshift, columnar storage is often more important than perfectly normalized tables because analysts usually care more about speed than perfect structure. Columnar storage means data is stored by column, not by row, so when an analyst runs a query, Redshift only reads the specific columns needed instead of the entire table. This makes queries much faster, especially when working with very large datasets. While normalization is great for transactional systems, analysts benefit more from quick scans and aggregations, even if that means some data is duplicated in less-normalized tables.

Question 2: Redshift's Leader Node and Compute Nodes work together to make reports generate faster. The Leader Node acts like a manager, by receiving the query and figuring out how to break it up. The Compute Nodes then do the heavy lifting by each working on a portion of the data at the same time. Once they’re done, the Leader Node puts all the results together and sends them back to the BI tool. Because the work is split across many nodes instead of handled by one, large reports and dashboards can be created much more quickly.
